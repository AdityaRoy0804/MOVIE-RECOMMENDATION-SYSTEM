### IMPORTING THE MODULES

```python
from surprise import SVD, Dataset, Reader
from surprise.model_selection import train_test_split
from surprise import accuracy
from collections import defaultdict
import pandas as pd
```

### LOADING THE DATA

```python
# Load the built-in MovieLens 100k dataset
data = Dataset.load_builtin('ml-100k')
```

### TRAIN-TEST SPLIT

```python
trainset, testset = train_test_split(data, test_size=0.2, random_state=42)
```

### TRAIN THE MODEL 

```python
model = SVD()
model.fit(trainset)
```

### MODEL EVALUATION

```python
predictions = model.test(testset)

# Evaluation metrics
print("RMSE:", accuracy.rmse(predictions))
print("MAE:", accuracy.mae(predictions))
```

### DEFINE FUNCTION FOR TOP-N RECOMMENDATIONS

```python
from collections import defaultdict

def get_top_n(predictions, n=5):
    top_n = defaultdict(list)
    for uid, iid, true_r, est, _ in predictions:
        top_n[uid].append((iid, est))
    
    # Sort predictions for each user
    for uid, user_ratings in top_n.items():
        user_ratings.sort(key=lambda x: x[1], reverse=True)
        top_n[uid] = user_ratings[:n]
    
    return top_n

top_n = get_top_n(predictions, n=3)


```

### LOAD MOVIE METADATA

```python
movie_path = r"C:\Users\DJ COMPUTERS\OneDrive\Desktop\Movie Recommender App\u.item"   # Adjust path if needed

movies = pd.read_csv(movie_path, sep='|', encoding='latin-1', header=None, usecols=[0, 1])
movies.columns = ['movie_id', 'title']
movie_id_to_title = dict(zip(movies['movie_id'].astype(str), movies['title']))
```

### DISPLAY RECOMMENDATIONS

```python
print("\n🎬 Top 3 Movie Recommendations per User:\n")
for uid, user_ratings in list(top_n.items())[:5]:  # show top 5 users
    print(f"User {uid}:")
    for iid, est_rating in user_ratings:
        movie_title = movie_id_to_title.get(iid, "Unknown Title")
        print(f"  📌 Movie ID: {iid}, 🎞️ Title: {movie_title}, ⭐ Predicted Rating: {round(est_rating, 2)}")
    print("-" * 60)
```

### SAVE THE MODEL
